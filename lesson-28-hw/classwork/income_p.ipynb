```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

df = pd.read_csv(r'G:\AI\AI\Homework AI\lesson-29\classwork\income.csv').drop(columns=['Name']).values
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(df_scaled)

y_kmeans = kmeans.labels_
print(y_kmeans)
centers = kmeans.cluster_centers_
print(centers)
plt.figure(figsize=(8, 6))

plt.scatter(df_scaled[:, 0], df_scaled[:, 1], c=y_kmeans, s=50, cmap='viridis', alpha=0.7)
plt.scatter(centers[:, 0], centers[:, 1], c='red', marker='*', s=200, label='Centroids', edgecolor='k')
plt.title('K-Means Clustering with scikit-learn')
plt.xlabel('Age')
plt.ylabel('Income')
plt.legend()
plt.show()
```
