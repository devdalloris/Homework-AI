```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

data = pd.read_csv(r'G:\AI\AI\Homework AI\lesson-29\classwork\income.csv').drop(columns=['Name']).values

scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)

kMeans = KMeans(n_clusters=3, random_state=42)
kMeans.fit(data_scaled)

y_means = kMeans.labels_
centers = kmeans.cluster_centers_

plt.figure(figsize=(8, 6))
plt.scatter(data_scaled[:, 0], data_scaled[:, 1], c=y_means, cmap='viridis', marker='*', s=50)
plt.title('K-Means Clustering of Income Data')
plt.xlabel('Age (scaled)')
plt.ylabel('Income (scaled)')
plt.show()
```
