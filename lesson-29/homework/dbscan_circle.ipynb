```python
from sklearn.datasets import make_circles
import matplotlib.pyplot as plt
X, y = make_circles(n_samples=900, factor=0.5, noise=0.05, random_state=42)
```

```python
plt.scatter(X[:,0], X[:,1])
plt.show()
```

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np
neigh = NearestNeighbors(n_neighbors=4)
neigh.fit(X)
distances, indices = neigh.kneighbors(X)
```

```python
distances = np.sort(distances[:, 1], axis=0)
```

```python
plt.figure(figsize=(6,4))
plt.plot(distances)
```

```python
from sklearn.cluster import KMeans, DBSCAN
kmeans = KMeans(n_clusters=2)
kmeans.fit(X)
clusters_kmeans = kmeans.predict(X)

dbscan = DBSCAN(eps = 0.1)
dbscan.fit(X)
clusters_db = dbscan.fit_predict(X)
```

```python
fig, axes = plt.subplots(ncols=2, nrows=1, figsize=(10,8))
plt.sca(axes[0])
plt.scatter(X[:,0], X[:,1], c=clusters_kmeans)
plt.sca(axes[1])
plt.scatter(X[:,0], X[:,1], c=clusters_db)
plt.show()
```

```python
from sklearn.cluster import DBSCAN

dbscan1 = DBSCAN(eps = 0.04)
dbscan1.fit(X)
clusters_db1 = dbscan1.fit_predict(X)

dbscan2 = DBSCAN(eps = 0.06)
dbscan2.fit(X)
clusters_db2 = dbscan2.fit_predict(X)

dbscan3 = DBSCAN(eps = 0.08)
dbscan3.fit(X)
clusters_db3 = dbscan3.fit_predict(X)

dbscan4 = DBSCAN(eps = 0.1)
dbscan4.fit(X)
clusters_db4 = dbscan4.fit_predict(X)
```

```python
fig, axes = plt.subplots(ncols=2, nrows=2, figsize=(8,7))
plt.sca(axes[0][0])
plt.scatter(X[:,0], X[:,1], c=clusters_db1)
plt.title("eps = 0.04")
plt.sca(axes[0][1])
plt.scatter(X[:,0], X[:,1], c=clusters_db2)
plt.title("eps = 0.06")
plt.sca(axes[1][0])
plt.scatter(X[:,0], X[:,1], c=clusters_db3)
plt.title("eps = 0.08")
plt.sca(axes[1][1])
plt.scatter(X[:,0], X[:,1], c=clusters_db4)
plt.title("eps = 0.1")
plt.show()
```
