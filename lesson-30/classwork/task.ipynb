```python
from sklearn.datasets import fetch_olivetti_faces
import matplotlib.pyplot as plt
X, _ = fetch_olivetti_faces(return_X_y=True)
```

```python
plt.imshow(X[0].reshape(64,64), cmap='gray')
plt.show()
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=80)
kmeans.fit(X)
y_means = kmeans.labels_
```

```python
mask = X[y_means==0]
```

```python
mask.shape
```

```python
def imshow(row):
    plt.imshow(row.reshape(64,64), cmap='gray')
    plt.axis('off')
```

```python
cluster_data = mask
cluster_data.shape[1]
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(8,8))
for i in range(2):
    for j in range(3):
        plt.sca(axes[i][j])
        imshow(cluster_data[i+j])
```
